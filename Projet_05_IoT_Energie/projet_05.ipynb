````python
# ==========================================
# QUÊTE 4 : Moteur d'Audit POO avec Dataframe Pandas
# ==========================================
# Consignes :
# 1. Parcourez chaque ligne du dataframe `df` (transmis en paramètre).
# 2. Calculez la puissance produite avec `self.panneau.calculer_puissance_kw(row['tension_v'], row['courant_a'])`.
# 3. Calculez l'énergie totale produite en kWh : somme des puissances × 0.25h (pas de 15 minutes).
# 4. Retournez un dictionnaire : {'energie_totale_kwh': round(energie_totale, 2), 'puissance_moyenne_kw': round(puissance_moyenne, 2)}.

# ==========================================
# QUÊTE 4 : Moteur d'Audit POO avec Dataframe Pandas
# ==========================================
# Consignes :
# 1. Parcourez chaque ligne du dataframe `df` (transmis en paramètre).
# 2. Calculez la puissance produite avec `self.panneau.calculer_puissance_kw(row['tension_v'], row['courant_a'])`.
# 3. Calculez l'énergie totale produite en kWh : somme des puissances × 0.25h (pas de 15 minutes).
# 4. Retournez un dictionnaire : {'energie_totale_kwh': round(energie_totale, 2), 'puissance_moyenne_kw': round(puissance_moyenne, 2)}.

import pandas as pd


class AuditEngine:

    def __init__(self, panneau: PanneauSolaire):
        self.panneau = panneau

    def auditer_dataframe(self, df: pd.DataFrame) -> dict:
        # ### START CODE HERE ###
        puissances = []
        for _, row in df.iterrows():
            puissance = self.panneau.calculer_puissance_kw(
                row["tension_v"], row["courant_a"]
            )
            puissances.append(puissance)

        # 3. Calcul de l'énergie totale (somme des puissances x 0.25h)
        energie_totale = sum(puissances) * 0.25

        # Calcul de la puissance moyenne
        puissance_moyenne = (
            sum(puissances) / len(puissances) if len(puissances) > 0 else 0.0
        )

        # 4. Retour du dictionnaire de résultats
        return {
            "energie_totale_kwh": round(energie_totale, 2),
            "puissance_moyenne_kw": round(puissance_moyenne, 2),
        }
        # ### END CODE HERE ###


# --- VALIDATION AUTOMATIQUE ---
df_telemetrie = pd.DataFrame(
    {
        "tension_v": [230.0, 228.0, 231.0, 229.0],
        "courant_a": [5.0, 10.0, 15.0, 20.0],
    }
)

auditeur = AuditEngine(panneau)
rapport_audit = auditeur.auditer_dataframe(df_telemetrie)

assert (
    rapport_audit["energie_totale_kwh"] == 2.87
), f"Attendu : 2.87 kWh, Obtenu : {rapport_audit['energie_totale_kwh']}"
assert (
    rapport_audit["puissance_moyenne_kw"] == 2.87
), f"Attendu : 2.87 kW, Obtenu : {rapport_audit['puissance_moyenne_kw']}"
print("🏆 Quête 4 réussie !")
````
