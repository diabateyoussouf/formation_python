## 💡 Le Concept de l'IoT (Internet des Objets)

Dans une centrale énergétique moderne, nous connectons des composants physiques à du code Python :

* **Le Capteur (Sensor) = Tes yeux** : Il mesure une grandeur physique (température en °C, pression en bar, débit d'eau en L/min) et la convertit en chiffre numérique.
* **Le Code Python = Ton cerveau** : Il lit ces chiffres, vérifie qu'il n'y a pas de danger, fait des calculs statistiques et prend des décisions.
* **L'Actionneur (Actuator) = Tes mains** : C'est un composant mécanique (pompe, vanne d'eau, ventilateur, alarme) déclenché par Python pour réagir.

## 🟡 Niveau 2 : L'Analyseur Temporel (Listes & Boucles `for`)

### 📚 Cours théorique pas-à-pas :

#### 1. C'est quoi une Liste ?
Un capteur ne prend pas une seule mesure : il prend une mesure toutes les minutes.
Créer une variable unique pour chaque mesure (`temp1`, `temp2`, `temp3`...) deviendrait vite impossible.
On utilise donc une **Liste**, un conteneur ordonné capable d'emballer des milliers de mesures :
```python
releves = [42.0, 45.5, 51.0, 58.0]

##### L'indexation (Attention : Python commence à 0 !) :

1. Premier élément : **`releves[0]`** ($\rightarrow 42.0$)
2. Deuxième élément : **`releves[1]`** ($\rightarrow 45.5$)
3. Nombre d'éléments : **`len(releves)`** ($\rightarrow 4$)

#### 2. C'est quoi une Boucle for et Pourquoi l'utiliser ?
Au lieu de **répéter** du code **1000** fois à la main, la boucle **`for`** ordonne à Python de parcourir automatiquement chaque élément d'une liste un par un pour lui appliquer une action.

```python
# Exemple de cumul (Somme progressive)
somme = 0
for temp in releves:
    somme = somme + temp  # On ajoute chaque valeur à la somme cumulée

#### 3. Logique de recherche du Maximum

On déclare un accumulateur pour garder en mémoire le plus grand nombre croisé jusqu'ici.

À chaque étape de la boucle, si la température courante est plus grande que notre maximum stocké, on met à jour notre maximum.

````python
s = 0
for i in range(11):
    s = s + i

print(f"La somme est :{s}")
print("La somme est:",s)
````

## 🟠 Niveau 3 : Le Nettoyeur de Signal (Filtrage de Données)

### 📚 Cours théorique pas-à-pas :

#### 1. Pourquoi le signal des capteurs est-il bruité ?
Sur le terrain, un capteur subit des baisses de tension ou de légères micro-coupures électriques. Quand cela se produit, le capteur renvoie souvent un chiffre absurde (ex: `-999.0` ou `9999.0`).
Si l'on calcule la moyenne avec ces fausses données, le résultat sera totalement défaillant !

#### 2. La Méthode du Filtrage par Liste Concourante
Pour éliminer les bugs :
1. On crée une **nouvelle liste vide** : `donnees_propres = []`
2. On parcourt la liste brute avec une boucle `for`.
3. Grâce à un test `if`, si la valeur est valide (différente de `-999.0`), on l'ajoute à la nouvelle liste avec la méthode **`.append(valeur)`**.

```text

## 📚 Fiche Réflexe : Le Moteur NumPy (À apprendre par cœur !)

En Python standard, une liste de listes `[[1, 2], [3, 4]]` est lente et peu pratique pour les calculs. **NumPy** introduit le type `array` (tableau matriciel), spécialement conçu pour le traitement de signal et la physique.

### Pourquoi utiliser NumPy au lieu des listes Python ?
Les listes Python sont très lentes dès qu'il s'agit de traiter des milliers de données physiques.
NumPy est la bibliothèque scientifique incontournable en ingénierie. Elle introduit l'objet ndarray (Tableau multidimensionnel) écrit en langage C, permettant de faire des calculs jusqu'à 100 fois plus rapides.

### 1. Attributs et Méthodes Essentiels

| Commande | Rôle Physique / Signification | Exemple | Résultat Obtenu |
|---|---|---|---|
| `np.array(liste)` | Convertit une liste standard en matrice optimisée | `A = np.array([[10, 20], [30, 40]])` | Objet `ndarray` 2D |
| **`.shape`** | Renvoie les dimensions sous la forme `(lignes, colonnes)` | `A.shape` | `(2, 2)` *(Attention : pas de `()` à la fin !)* |
| **`.max()`** | Trouve la valeur pic absolue sur toute la grille | `A.max()` | `40` |
| **`.min()`** | Trouve la valeur la plus basse du tableau | `A.min()` | `10` |
| **`.mean()`** | Calcule la température moyenne globale | `A.mean()` | `25.0` |
| **`.sum()`** | Calcule la somme de toutes les cases | `A.sum()` | `100` |

---

### 2. Le Piège des Axes : `axis=0` vs `axis=1`

Quand on applique une fonction statistique (`.mean()`, `.max()`, `.sum()`) sur une matrice, on doit choisir la direction de calcul :

* **`axis=0` (Axe Vertical / Colonnes)** : Analyse **capteur par capteur**. On écrase les lignes pour obtenir une valeur par colonne.
  * *Exemple IoT* : `matrice.mean(axis=0)` $\rightarrow$ Donne la température moyenne de chaque capteur sur la journée.
* **`axis=1` (Axe Horizontal / Lignes)** : Analyse **instant par instant**. On écrase les colonnes pour obtenir une valeur par tranche horaire.
  * *Exemple IoT* : `matrice.mean(axis=1)` $\rightarrow$ Donne la température moyenne globale du champ à l'instant $t$.

```text
               axis=0 (Vertical)
                   ↓   ↓   ↓
Ligne 0 (t=0h)  [ 20 , 30 , 40 ]  ──> axis=1 (Horizontal)
Ligne 1 (t=1h)  [ 50 , 60 , 70 ]  ──> axis=1 (Horizontal)

#### 5. C'est quoi le Slicing [:, :] et Pourquoi l'utiliser ?
Au lieu d'écrire de longues boucles for pour extraire des éléments, le Slicing permet de découper une matrice en une seule ligne de code !
Syntaxe : **`matrice[lignes, colonnes]`** (Le symbole deux-points : signifie "PRENDRE TOUT").

**`matrice[:, 1]`** : Prend TOUTES les lignes de la `Colonne 1` (Historique du Capteur B).

**`matrice[2, :]`** : Prend la Ligne 2 complète sur TOUTES les `colonnes` (Relevé global à t=2h).

````python
matrice = np.array([
    [1,1,3],
    [1,1,2],
    [1,1,5]
])
````

#### 2. Le Filtre à Moyenne Glissante 1D (Lissage de Signal)
Les oscillations parasites de haute fréquence altèrent la lisibilité des signaux.
Le **filtre à moyenne glissante 3 points** remplace la valeur d'un point par la moyenne calculée entre lui-même, son voisin de gauche et son voisin de droite :

$$\text{Valeur Lissée}[i] = \frac{\text{Signal}[i-1] + \text{Signal}[i] + \text{Signal}[i+1]}{3}$$

*Remarque : Nous commençons la boucle à l'index `1` et la terminons à `len(signal) - 1` car le premier et le dernier point n'ont pas assez de voisins !*
